### DCR Graph as the process model

The DCR from the paper. This not stable because I can make a quality that was true false again. But it is 
weakly compliant, as I can always satisfy the quality.

![DCR from paper](images/example/dcr_paper.jpg)

```python

goal_model = read_istar_model("Data/example/goal_model.txt")
dcr = read_dcr("Data/example/dcr_paper.xml")
event_mapping = read_event_mapping_csv("Data/example/dcr_map.csv")

analyse_models(goal_model, dcr, event_mapping)
```

To understand what is going on, one can open the DCR simulator on or DCR Graphs https://hugoalopez-dtu.github.io/dcr-js/ (read the file as a DCR Solutions XML) and execute the traces given as counter examples and also run the What-if interactive animation for the goal model.

```python
interface = WhatIfInterfaceBuilder(goal_model).create_interface()
display(interface)
    
```

This is an alternate version, where one cannot repeat the process. In this case, it is stable, because either we end in a satisfied quality and we are done. Or the quality is not satisfied, but then we are also done.

The example, however, is not weakly compliant, as we can always get to a state where the quality is not satisfied and we have reached the end of the execution.

![DCR from paper fixed](images/example/dcr_fixed.jpg)

```python

goal_model = read_istar_model("Data/example/goal_model.txt")
dcr = read_dcr("Data/example/dcr_fixed.xml")
event_mapping = read_event_mapping_csv("Data/example/dcr_map.csv")
analyse_models(goal_model, dcr, event_mapping)
```

### Failing Weak compliance in the security example
Weak compliance can fail, if an intentional element of the goal model is not mapped to a transition in the event mapping.

```python

goal_model = read_istar_model("Data/security/goal_model.txt")
petri_net = read_petri_net("Data/security/petri_net.pnml")

event_mapping = petri_net.get_default_event_mapping()
event_mapping.update({"t_6":[]}) # remove the mapping for t6 to show failing weak compliance

analyse_models(goal_model, petri_net, event_mapping=event_mapping)

interface = InterfaceBuilder(goal_model, petri_net, event_mapping=event_mapping).create_interface()
display(interface)
```
