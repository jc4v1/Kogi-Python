### What-If Scenario

The non-interactive version of the What-If scenario creates for a goal model a Petri net, which consists of one place and for each of the leaf intensional elements of the goal model a transition. When running "analyse_models", stability and weak compliance is checked, as well as the number of the reachable states and transitions for the combined system is shown.

Note that in the example below (the goal model for the running example), the number of reachable states are the same as that of the goal model, but the number of transitions is less. The reason for this is, that the combined system only executes leaf intensional elements, which results in only 4 actions, while in the original goal model has 8 intensional events without the quality, thus 8 actions. 

```python

goal_model = read_istar_model("Data/example/goal_model.txt")
print(f"Goal Model LTS Actions: {goal_model.as_transition_system().actions()} ")
process_model = goal_model.generate_all_events_petri_net()
event_mapping = { e: [[e]] for e in goal_model.get_events()}

analyse_models(goal_model, process_model, event_mapping)
```

To start the interactive version of the What-If scenario, you can execute.

```python
goal_model = read_istar_model("Data/example/goal_model.txt")

interface = WhatIfInterfaceBuilder(goal_model).create_interface()
display(interface)
```

### Fail: Not Weakly Compliant


Here, the Petri Net is updated according to the description in the paper by removing the transition for t10, which was mapped to "Record the@ justification for@ exclusion as an@ extraordinary circumstanc". This leads to a system that is not weakly compliant because the quality "Extraordinary@ circumstances@ documented@ appropriately" is not satisfied. The non-interactive mode generates counter example traces for this, leading to a state, where the quality can't be satisfied anymore. The interactive scenario allows to select such a state and jump directly to that state. 


```python

goal_model = read_istar_model("Data/air-passenger-rights/goal_model.txt")
petri_net = read_petri_net("Data/air-passenger-rights/petri_net_fail.pnml")
event_mapping = petri_net.get_default_event_mapping()

analyse_models(goal_model, petri_net, event_mapping)

interface = InterfaceBuilder(goal_model
                             , petri_net
                             , event_mapping=petri_net.get_default_event_mapping()).create_interface()

display(interface)
```

Recommended BPMN editor <a href="https://bpmn.io/">BPMN.io</a>



Kogi does not have a direct implementation of the semantics of BPMN models. Instead BPMN models are converted
to Petri nets based on the pm4py library. 

However, the Petri net being created by pm4py is not directly suited for Kogi. However, the Petri net can be converted
to a Petri net useful for Kogi using the method convert_bpmn_net and providing a mapping.

```python
import pm4py
goal_model = read_istar_model('Data/bpmn/goal_model.txt')
bpmn_model = pm4py.read_bpmn("Data/bpmn/process_model.bpmn")
mapping = read_event_mapping_csv("Data/bpmn/map.csv")

net, initial_marking, final_marking = pm4py.convert_to_petri_net(bpmn_model)
petri_net = PetriNet(net,initial_marking,final_marking,{})
petri_net.convert_bpmn_net(mapping)

analyse_models(goal_model, petri_net, mapping)

display(WhatIfInterfaceBuilder(goal_model).create_interface())
```
