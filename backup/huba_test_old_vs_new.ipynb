# Comparison between the new and old semantics
- ``gm.kogi=True`` sets the old semantics
- ``gm.kogi=False`` or the default semantics is the new semantics.

The example with dependencies.

```python
from Semantics.istar_processor import read_istar_model
from Ui.interface import SingleStepInterfaceBuilder, WhatIfInterfaceBuilder

```

```python

goal_model_path = "content/test/test0e_fail.txt"

goal_model_kogi = read_istar_model(str(goal_model_path), qualified=True)
goal_model_kogi.kogi = True

interface = WhatIfInterfaceBuilder(goal_model_kogi).create_interface()
display(interface)

goal_model_new = read_istar_model(str(goal_model_path), qualified=True)
goal_model_new.kogi = False
interface = WhatIfInterfaceBuilder(goal_model_new).create_interface()
display(interface)
```

# New Semantics, Single Step, Kogi Semantics, Single Step

Using the ``SingleStepInterfaceBuilder`` gives one a ``WhatIfInterfaceBuilder`` but with executing only single steps, showing which parents are affected and which rule has been applied, if any.

This can be used for debugging the rules.

The following cell contains all 4 combinations: new semantics what-if scenario multistep and single step and kogi semantics what-if scenario multipstep and single step.

```python
goal_model_path = "content/gm_bp_dep.txt"

goal_model_new = read_istar_model(str(goal_model_path))
goal_model_new.kogi = False

interface = WhatIfInterfaceBuilder(goal_model_new).create_interface()
display(interface)

goal_model_single_step = read_istar_model(str(goal_model_path))
goal_model_single_step.kogi = False

interface = SingleStepInterfaceBuilder(goal_model_single_step).create_interface()
display(interface)

goal_model_kogi = read_istar_model(str(goal_model_path))
goal_model_kogi.kogi = True

interface = WhatIfInterfaceBuilder(goal_model_kogi).create_interface()
display(interface)

goal_model_kogi_single_step = read_istar_model(str(goal_model_path))
goal_model_kogi_single_step.kogi = True

interface = SingleStepInterfaceBuilder(goal_model_kogi_single_step).create_interface()
display(interface)

```
